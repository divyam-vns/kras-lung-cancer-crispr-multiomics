```python
download_script = r'''
import os
import requests

os.makedirs("data/raw", exist_ok=True)

files = {
    "CRISPRGeneEffect.csv":
    "https://depmap.org/portal/api/download/gene_dependency",

    "Model.csv":
    "https://depmap.org/portal/api/download/all_models"
}

for filename, url in files.items():

    print(f"Downloading {filename}...")

    response = requests.get(url)

    with open(f"data/raw/{filename}", "wb") as f:
        f.write(response.content)

    print(f"Saved: data/raw/{filename}")

print("Download complete.")
'''

with open("scripts_download_depmap.py", "w") as f:
    f.write(download_script)

print("Script saved.")
```

```python
import shutil
import os

backup_dir = "/content/drive/MyDrive/kras_lung_cancer_project"

folders_to_backup = [
    "data",
    "results",
    "figures"
]

files_to_backup = [
    "scripts_load_data.py"
]

for folder in folders_to_backup:

    if os.path.exists(folder):

        shutil.copytree(
            folder,
            os.path.join(backup_dir, folder),
            dirs_exist_ok=True
        )

for file in files_to_backup:

    if os.path.exists(file):

        shutil.copy(
            file,
            backup_dir
        )

print("Progress saved.")
```

```python
mutation_script = r'''
import pandas as pd

models = pd.read_csv(
    "data/raw/Model.csv"
)

mutations = pd.read_csv(
    "data/raw/OmicsSomaticMutations.csv",
    low_memory=False
)

kras = mutations[
    mutations["HugoSymbol"] == "KRAS"
]

lung_models = models[
    models["OncotreeLineage"] == "Lung"
]

print("KRAS mutations:", kras.shape)
print("Lung models:", lung_models.shape)
'''

with open(
    "scripts_kras_mutations.py",
    "w"
) as f:
    f.write(mutation_script)

print("Script saved.")
```

```python
kras_model_ids = kras["ModelID"].unique()

print("KRAS mutant models:", len(kras_model_ids))
print(kras_model_ids[:10])
```

```python
kras_lung_models = lung_models[
    lung_models["ModelID"].isin(kras_model_ids)
]

print("KRAS-mutant lung cancer models:", kras_lung_models.shape)
kras_lung_models[["ModelID","CellLineName","OncotreePrimaryDisease"]].head()
```

```python
wt_lung_models = lung_models[
    ~lung_models["ModelID"].isin(kras_model_ids)
]

print("KRAS wild-type lung cancer models:", wt_lung_models.shape)
wt_lung_models[["ModelID","CellLineName"]].head()
```
